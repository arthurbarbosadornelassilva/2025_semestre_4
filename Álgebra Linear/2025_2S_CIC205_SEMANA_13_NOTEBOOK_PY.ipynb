## PARTE 2: Produto Vetorial 

### O que é?
Dados dois vetores $\vec{a}$ e $\vec{b}$, o produto vetorial $\vec{c} = \vec{a} \times \vec{b}$ resulta em um vetor **perpendicular** a ambos.

### Fórmula:
$$\vec{a} \times \vec{b} = \begin{vmatrix} \mathbf{i} & \mathbf{j} & \mathbf{k} \\ a_1 & a_2 & a_3 \\ b_1 & b_2 & b_3 \end{vmatrix}$$

### Em Python:
```python
c = np.cross(a, b)
```

## PARTE 3: Produto Escalar 

### O que é?
O produto escalar de $\vec{a}$ e $\vec{b}$ é um **número** (não um vetor):

$$\vec{a} \cdot \vec{b} = a_1b_1 + a_2b_2 + a_3b_3 = |\vec{a}||\vec{b}|\cos(\theta)$$

onde $\theta$ é o ângulo entre os vetores.

### Em Python:
```python
produto = np.dot(a, b)
```

## PARTE 4: Similaridade de Cosseno 

### Conceito:
A **similaridade de cosseno** mede quão "parecidos" são dois vetores:

$$\text{sim}(\vec{a}, \vec{b}) = \frac{\vec{a} \cdot \vec{b}}{|\vec{a}||\vec{b}|} = \cos(\theta)$$

- `sim = 1`: vetores idênticos (ângulo 0°)
- `sim = 0`: vetores perpendiculares (ângulo 90°)
- `sim = -1`: vetores opostos (ângulo 180°)
